```python
#calculator program
var1=int(input("Select the operation\n1.Addition\n2.Subtraction\n3.Multiplication\n4.Division"))
if(var1==1):
  num1=float(input("Enter the first number"))
  num2=float(input("Enter the second number"))
  sum=num1+num2
  print(sum)
elif(var1==2):
  num1=float(input("Enter the first number"))
  num2=float(input("Enter the second number"))
  diff=num1-num2
  print(diff)
elif(var1==3):
  num1=float(input("Enter the first number"))
  num2=float(input("Enter the second number"))
  product=num1*num2
  print(product)
elif(var1==4):
  num1=float(input("Enter the first number"))
  num2=float(input("Enter the second number"))
  if(num2==0):
    print("Division with zero is not possible")
  else:
    quo=num1/num2
    print(quo)
else:
  print("Entry not valid")




```
