```python
#Number guessing game
import random
num=random.randint(1,100)
print("Enter your guess between 1 to 100")
while True:
  guess=int(input())
  if(guess<num):
    print("Try higher")
  elif(guess>num):
    print("Try lower")
  elif(guess==num):
    print("Your guess is correct!!")
  else:
    print("Your guess must be between 1 to 100")

```
